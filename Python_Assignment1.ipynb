Q1. Write a function that takes a username string as input and validates it. To be valid, the username must be at
least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and
numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase
version of the username. If it is invalid, return the string "Invalid".

```python
str=input()

def validation(str):
    str=str.strip()
    if(len(str)<5):
        return "Invalid"
    
    if(str.isalnum()==False):
        return "Invalid"
    
    return str.lower()

print(validation(str))

```

Q2 You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the
order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

Eg: secret_code = "charlie-bravo-alpha" "ALPHA BRAVO CHARLIE"

```python
str=input()

def decode(str):
    str=str.strip("- ")

    arr=str.split("-");
    arr=arr[::-1]

    for i in range(len(arr)):
        arr[i]=arr[i].upper()

    ans=" ".join(arr)
    return ans

print(decode(str))
```

You have received a raw string of names from a database exported as a semicolon-separated list. The data is
messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every
name is properly Title Cased, extra spaces are removed, and the names are joined into a single string
separated by a comma and a space.

```python
str=input()

def cleanData(str):
    str=str.strip("; ")

    arr=str.split(";")
    for i in range(len(arr)):
        arr[i]=arr[i].title()

    ans=", ".join(arr)
    return ans

print(cleanData(str))

```

Q4. A user has accidentally typed a sentence with multiple random spaces between words, and the casing is entirely
lowercase. Write code that removes all the extra spaces so there is exactly one space between each word, and
capitalizes only the very first letter of the entire sentence.

```python
str=input()

def cleanSentence(str):
    str=str.strip()
    arr=str.split()

    ans=" ".join(arr).capitalize()
    return ans

print(cleanSentence(str))
```

Q5. Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces
and non-alphanumeric characters, convert everything to lowercase, and check whether it reads the same
forwards and backwards.

```python
str=input()

def isPalindrome(str):
    str=str.strip()
    arr=[]

    for ch in str:
        if(ch.isalnum()):
            arr.append(ch)

    sen="".join(arr).lower()

    return sen==sen[::-1]

print(isPalindrome(str))
```

Q6. Given a string representing a social media post, extract all the hashtags. The extracted hashtags must be
completely lowercase, and the # symbol should be removed. Store them in a list. Ignore any words that are not
hashtags.

```python
str=input()

def socialMedia(str):
    str=str.strip()
    arr=str.split()
    a=[]

    for i in range(len(arr)):
        if(arr[i].startswith("#")):
            a.append(arr[i])

    for i in range(len(a)):
        a[i]=a[i].strip("#").lower()

    return a

print(socialMedia(str))
```

Q7. You are given a single comma-separated string containing various filenames. Extract just the file extensions
(without the dot), ensure there are no duplicates, and return them as a list of strings strictly in UPPERCASE

```python
str=input()
def extensions(str):
    str=str.strip()
    arr=str.split(",")

    for i in range(len(arr)):
        arr[i]=arr[i].strip()
        idx=0
        for ch in range(len(arr[i])):
            if(arr[i][ch]=='.'):
                idx=ch+1
        arr[i]=arr[i][idx:]
        
    ans=[]
    s=set()

    for i in arr:
        if(i not in s):
            s.add(i)
            ans.append(i.upper())

    return ans

print(extensions(str))
```

Q8. Given an email address string, extract just the company/domain name — the part immediately after the @
symbol and before the dot suffix like .com or .edu. Convert the extracted domain to Title Case. If the string
does not contain an @, return "Invalid".

```python
str=input()
def findCompanyName(str):
    str=str.strip()

    if("@" not in str):
        return "Invalid"

    domainName=str.split("@")[1]
    companyName=domainName.split(".")[0]

    return companyName

print(findCompanyName(str))
```

Q9. Write a function to check if two given strings are anagrams of each other — they contain the exact same letters
in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as
the same.

```python
str1=input()
str2=input()

def isAnagram(str1, str2):
    str1=str1.strip().lower()
    str2=str2.strip().lower()

    hm={}

    for i in str1:
        if(i==' '):
            continue
        else:
            hm[i]=hm.get(i,0)+1

    for i in str2:
        if(i==' '):
            continue

        if(i in hm):
            if(hm.get(i)==0):
                return False
            else:
                hm[i]=hm.get(i)-1
        else:
            return False

    return True

print("Is Anagram: ",isAnagram(str1,str2))
```

Q10. You are building a CRM system and user-submitted phone numbers are a mess — they contain spaces,
parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by
removing all non-digit characters. Once cleaned, validate the number:
• If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".
• If it contains anything other than 10 digits, return "Invalid Number".


```python
num=input()

def phoneNum(num):
    num=num.strip()
    arr=[]

    for i in num:
        if(i.isdigit()==True):
            arr.append(i)

    if(len(arr)!=10):
        return "Invalid Number"

    s="("+"".join(arr[0:3])+")"+" "+"".join(arr[3:6])+"-"+"".join(arr[6:])
    return s

print(phoneNum(num))

```

Q11. An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly
3 uppercase letters, a single hyphen, and exactly 4 digits (e.g. "ABC-1234").
Write a function that receives a messy SKU string. First clean it by stripping leading/trailing spaces and
converting it to uppercase. Then validate it using string methods. Return "Valid: [Cleaned_SKU]" if it
passes, or "Invalid SKU" if it fails any formatting rule.

```python
code=input()
def validCode(code):
    code=code.strip()

    if("-" not in code):
        return "Invalid SKU";

    code=code.upper()
    alpha=code.split("-")[0]

    if(alpha.isalpha()==False or len(alpha)!=3):
        return "Invalid SKU"

    num=code.split("-")[1]
    if(num.isdigit()==False or len(num)!=4):
        return "Invalid SKU"

    ans="Valid: "+alpha+"-"+num

    return ans

print(validCode(code))
```
