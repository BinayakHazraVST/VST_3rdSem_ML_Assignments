```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
bikeResale=pd.read_csv("./bike_resale.csv")
```

```python
bikeResale.info()
```

```python
bikeResale.describe().columns
```

```python
nums_col=['age_months', 'km_driven', 'owner_count', 'mileage_kmpl', 'service_count'];
cat_col=['brand','condition','insurance']
```

```python
x=bikeResale.drop(columns=['bike_id', 'price_thousand', 'sold_in_week']);
y=bikeResale['sold_in_week']
```

```python
for col in nums_col:
    x[col]=x[col].fillna(x[col].mean())
```

```python
print("Missing data in x:", x.isna().sum().sum())
print("Missing data in y:", y.isna().sum().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer

model=ColumnTransformer([
    ("StringConversion", OneHotEncoder(sparse_output=False), cat_col),
    ("Numeric_Standardization", StandardScaler(), nums_col)
])

x_final=model.fit_transform(x)
print(x_final)
```
