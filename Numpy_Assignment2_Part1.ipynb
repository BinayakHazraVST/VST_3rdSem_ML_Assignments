Q7. The result summary
Tasks

Print the class average and the middle value of the class.
Print the gap between the highest and the lowest marks.
Print the index positions of the topper and of the weakest student.
Print the second highest marks in the class.
Print how many students scored above the class average.
Print how many students lie within one standard deviation of the class average.
Print the class average after removing the topper.

```python
marks = np.array([72, 45, 88, 95, 30, 67, 91, 54, 78, 39])
```

```python
avg=np.mean(marks)
median=np.median(marks)
print(avg,median)

maxMark=np.max(marks)
minMark=np.min(marks)
print(maxMark-minMark)

topperPos=np.where(marks==maxMark)
print(topperPos[0][0])
weakPos=np.where(marks==minMark)
print(weakPos[0][0])

secondHigh=np.sort(marks)[-2]
print(secondHigh)

student1=marks[marks>avg]
print(len(student1))

std=np.std(marks)
lower=avg-std
high=avg+std
student2=marks[(marks>=lower) & (marks<=high)]
print(len(student2))


student3=marks[marks!=maxMark]
print(np.round(np.mean(student3),2))
```

```python
marks = np.array([[72, 85, 60, 91],
                  [45, 30, 98, 55],
                  [88, 66, 74, 100],
                  [95, 92, 55, 40],
                  [30, 48, 62, 70]])
```

```python
subject_averages = np.mean(marks, axis=0)
student_totals = np.sum(marks, axis=1)
print("Task 1:")
print(f"Average of each subject: {subject_averages}")
print(f"Total of each student: {student_totals}\n")

toughest_subject_idx = np.argmin(subject_averages)
print("Task 2:")
print(f"Index of toughest subject: {toughest_subject_idx}\n")


overall_topper_idx = np.argmax(student_totals)
print("Task 3:")
print(f"Index of overall topper: {overall_topper_idx}\n")


subject_toppers_marks = np.max(marks, axis=0)
print("Task 4:")
print(f"Highest marks in each subject: {subject_toppers_marks}\n")

above_60_count = np.sum(marks > 60, axis=0)
print("Task 5:")
print(f"Students scoring above 60 per subject: {above_60_count}\n")


most_spread_subject_idx = np.argmax(np.std(marks, axis=0))
print("Task 6:")
print(f"Index of most spread out subject: {most_spread_subject_idx}\n")

chemistry_idx = 2 
lowest_chem_student_idx = np.argmin(marks[:, chemistry_idx])
lowest_chem_student_total = student_totals[lowest_chem_student_idx]
print("Task 7:")
print(f"Total marks of student lowest in Chemistry (assuming index {chemistry_idx}): {lowest_chem_student_total}")
```


Q10. Grace marks and weightage
Tasks

Apply the subject-wise grace marks to the whole class, keeping every value at or below 100.
Multiply each subject by its own weightage and print the table rounded to 1 decimal.
Produce a table showing how far each mark sits from that subject's own class average.
Give 5 bonus marks to the first student, 0 to the second, 2 to the third and 0 to the fourth — in a single operation.
From the table of task 3, print how many entries lie below the subject average.
Produce a table where every subject is brought to one scale by subtracting its average and dividing by its spread. Round to 2 decimals.

```python
marks = np.array([[72, 85, 60, 91],
                  [45, 30, 98, 55],
                  [88, 66, 74, 100],
                  [95, 92, 55, 40]])

grace     = np.array([5, 0, 8, 3])
weightage = np.array([1.2, 1.0, 1.1, 1.0])
```

```python
marks_with_grace = np.minimum(marks + grace, 100)
print("Task 1:")
print(f"{marks_with_grace}\n")

weighted_marks = np.round(marks * weightage, 1)
print("Task 2:")
print(f"{weighted_marks}\n")

subject_averages = np.mean(marks, axis=0)
distance_from_avg = marks - subject_averages
print("Task 3:")
print(f"{distance_from_avg}\n")

bonus = np.array([5, 0, 2, 0])

marks_with_bonus = marks + bonus.reshape(-1, 1)
print("Task 4:")
print(f"{marks_with_bonus}\n")

below_avg_count = np.sum(distance_from_avg < 0)
print("Task 5:")
print(f"{below_avg_count}\n")

subject_spread = np.std(marks, axis=0)
scaled_marks = np.round((marks - subject_averages) / subject_spread, 2)
print("Task 6:")
print(f"{scaled_marks}")
```
