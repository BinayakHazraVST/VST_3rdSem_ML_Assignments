A5.
The owner wants to flag the generous customers. Make a column called service. Everyone starts as 'normal'.
Any bill where tip_percent is more than 20 becomes 'generous'. Then count how many of each. (Uses the
column you made in A1.)

B2.
Two counts:

```python
#1
print(((t['Sex']=='female') & (t['Pclass']==3)).sum())

#2
print(((t['Sex']=='female') & (t['Pclass']==3) & (t['Survived']==1)).sum())
```

B3.
Some people paid an enormous amount for their ticket. Show only Name, Age and Fare for every passenger who

```python
print(t.loc[t['Fare']>200, ['Name','Age','Fare']])
```

B4.
Out of all the passengers in first class, what percentage survived? Now do the same for third class. Put the two
numbers next to each other in a comment.

```python
firstClass=(t['Pclass']==1).sum()
survivedFirst=((t['Pclass']==1) & (t['Survived']==1)).sum()
percentSurvivedFirst=(survivedFirst/firstClass*100).round(1)

thirdClass=(t['Pclass']==3).sum()
survivedThird=((t['Pclass']==3) & (t['Survived']==1)).sum()
percentSurvivedThird=(survivedThird/thirdClass*100).round(1)

print("Survived 1st class", percentSurvivedFirst)
print("Survived 3rd class", percentSurvivedThird)
```
