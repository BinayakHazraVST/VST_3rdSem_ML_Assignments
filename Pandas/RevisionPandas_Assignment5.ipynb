Q3. Filter rows

How many cars manage more than 30 mpg? Of those, how many are
Japanese?

```python
carsGreater30mpg=(mpg['mpg']>30).sum()
print("Cars with mpg greater than 30:",carsGreater30mpg)

japaneseCarGreater30mpg=((mpg['mpg']>30) & (mpg['origin']=='japan')).sum()
print("Japanese Cars with mpg greater than 30:",japaneseCarGreater30mpg)
```

Q4. Handle missing values
horsepower has blanks.

1. Which origins do the missing rows belong to?
2. Fill the blanks with the median horsepower.
3. Confirm there are no blanks left.

```python
originsWithMissingHorsepower= mpg.loc[mpg['horsepower'].isnull(), 'origin']
print("Origin with Missing Horsepower:\n",originsWithMissingHorsepower)
```

```python
fillHorsepower=mpg['horsepower'].fillna((mpg['horsepower'].median()).round(2))
mpg['horsepower']=fillHorsepower
print("Filled Horsepower:\n",fillHorsepower)
```

```python
countNull=(mpg.isnull().sum()).sum()
print("Number of Null Data:",countNull)
```
