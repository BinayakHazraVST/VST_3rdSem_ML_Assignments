Q4. Handle missing values
horsepower has blanks.

1. Which origins do the missing rows belong to?
2. Fill the blanks with the median horsepower.
3. Confirm there are no blanks left.

```python
originsWithMissingHorsepower= mpg.loc[mpg['horsepower'].isnull(), 'origin']
print("Origin with Missing Horsepower:\n",originsWithMissingHorsepower)
```

Q5. Create a new column

Power-to-weight ratio is a better predictor of how a car feels than horsepower
alone.
Add a column power_to_weight = horsepower ÷ weight. Then show the 5
cars with the highest ratio.

```python
mpg['power_to_weight']=(mpg['horsepower']/mpg['weight']).round(2)
carsWithHighestRatio=mpg.sort_values(by='power_to_weight', ascending=False)
print("The first 5 cars with highest ratio:\n",carsWithHighestRatio.head())
```

Q7. Named aggregation

Build one table with a row per origin and these columns: cars , avg_mpg ,
avg_weight , avg_cylinders . Round to 1 decimal.

```python
carDetails=mpg.groupby("origin").agg(
    cars=("name","size"),
    avg_mpg=("mpg","mean"),
    avg_weight=("weight","mean"),
    avg_cylinders=("cylinders","mean")
).round(1)

print("Table of car details:\n",carDetails)
```

Q9. Group by two columns

Average mpg for every combination of origin and cylinders . Then
reshape it with .unstack() so origins are rows and cylinder counts are
columns.

```python
avgMsg=(mpg.groupby(['origin','cylinders'])['mpg'].mean()).round(2)
print("Average msg for different combination of origin and cylinders:",avgMsg)

unstackedTable=avgMsg.unstack()
print("The unstack table:\n",unstackedTable)
```

Q14. Compound filter

Find American cars that are 4-cylinder, from model year 1978 or later, and
manage more than 25 mpg. How many are there? Show the 5 most efficient.

Q15. String methods

The name column holds make and model together, like chevrolet chevelle
malibu .
1. Extract the brand — the first word — into a new brand column.
2. List every distinct brand.
3. Something is wrong with that list. Find the problem and fix it.

```python
mpg['brand']=mpg['name'].str.strip().str.split()
mpg['brand']=mpg['brand'].str[0]
print(mpg['brand'])
```

```python
print("Distinct Brand:\n", mpg['brand'].unique())
```

```python
spellingCorrection={
    'chevy': 'chevrolet',
    'chevroelt': 'chevrolet',
    'toyouta': 'toyota',
    'maxda': 'mazda',
    'vokswagen': 'volkswagen',
    'vw': 'volkswagen',
    'mercedes': 'mercedes-benz',
    'hi': 'ihc'
}

mpg['brand']=mpg['brand'].replace(spellingCorrection)

print("Corrected Brand Spelling:\n",mpg['brand'].unique())
```

Q16. Brand summary

Using your cleaned brand column: for every brand with at least 15 cars,
show the number of cars and the average mpg, sorted from most to least efficient.

```python
carEfficiencyDetails=mpg.groupby("brand").agg(
    car_number=("mpg","count"),
    avg_mpg=("mpg","mean"),
).sort_values(by="avg_mpg").round(2)

print("The car filter brand:\n", carEfficiencyDetails.loc[carEfficiencyDetails["car_number"]>=15,:])
```

Q17. Change over time

1. What was the average mpg of the 1970 fleet? The 1982 fleet?
2. Someone concludes: "Japanese cars got much more efficient over the
decade — that is why they took the market." Check it. Did Japanese cars
improve more than everyone else's?

```python
avgMpg70=mpg.loc[mpg['model_year']==70,['mpg']].mean().round(2)
print("Average mpg of 1970 fleet:",avgMpg70)

avgMpg82=mpg.loc[mpg['model_year']==82,['mpg']].mean().round(2)
print("Average mpg of 1982 fleet:",avgMpg82)
```

```python
sortYearData=mpg.sort_values(by="model_year")
compareMpgMedian=sortYearData.groupby("origin")['mpg'].median().round(2)
compareMpgMean=sortYearData.groupby("origin")['mpg'].mean().round(2)

print("The comparison of MPG based on median:\n",compareMpgMedian)
print("The comparison of MPG based on mean:\n",compareMpgMean)

# As the median and mean of mpg for japan sorted based on year is more than any other => hence, it has evolved over time and the statement is true
```
