3. Find the number of passengers in each passenger class using both
count() and size().

```python
countPassenger=titanic.groupby("Pclass")['Pclass'].count()
sizePassenger=titanic.groupby("Pclass").size()

print("Count of Passenger\n",countPassenger)
print("Size of Passenger\n",sizePassenger)
```
