```python
import numpy as np;
import pandas as pd;
```

1. Create a DataFrame containing 8 students with columns such as Name, Marks,
Age, and City. Include some missing values.

```python
studentDetails={
    "Name":["Binayak", "Debnil", "Virat", "Rohit", "Raju", "Rahul", "Aditya", "Rohan"],

    "Marks":[100, 90, 60, 80, np.nan , 70, np.nan , 90],

    "Age": [18, 20, np.nan, 35, 30, 25, np.nan, 21],

    "City": ['Kolkata', 'Bengalore', None, "Mumbai", "Chennai", "Hyderabad", None, "Indore"],
}

students=pd.DataFrame(studentDetails)
studentsCopy=students
print(students);
```

2. Find the number of missing values in each column using isnull()

```python
missingValNumber=students.isnull().sum();
print("The number of missing values: \n",missingValNumber)
```

3. Display all rows that contain at least one missing value.

```python
rowsWithMissingValue=students[students.isnull().any(axis=1)]
print("Rows with Missing values: \n",rowsWithMissingValue)
```

4. Remove all rows containing missing values using dropna().

```python
cleanStudent=students.dropna()
print("Cleaned Student details:\n",cleanStudent)
```

5. Remove only the rows where the Marks column contains missing values.

```python
cleanStudentCol=students.dropna(axis=1)
print("Cleaned Student Columns\n",cleanStudentCol)
```

6. Fill missing values in the Marks column using the mean of the column.

```python
filledMarks=students["Marks"].fillna((students["Marks"].mean()).round(2))
students["Marks"]=filledMarks
print("Filled Marks: \n",filledMarks)
print("\n",students)
```

7. Fill missing values in the Age column using the median of the column.

```python
filledAge=students['Age'].fillna((students["Age"].median()).round(2))
students['Age']=filledAge
print("Filled Age:\n",filledAge)
```

8. Replace missing values in the City column with "Unknown".

```python
filledCity=students['City'].fillna("Unknown")
students['City']=filledCity
print("Filled City:\n",filledCity)
```

9. Create a DataFrame containing duplicate records and identify the duplicate rows
using duplicated().

```python
data = {
    "StudentID": [101, 102, 103, 102, 101],
    "Name": ["Aarav", "Priya", "Rohan", "Priya", "Aarav"],
    "Marks": [85, 90, 78, 90, 89],  # Row 1 and 3 are 100% duplicates across all columns
    "City": ["Mumbai", "Delhi", "Pune", "Delhi", "Mumbai"],
}

df=pd.DataFrame(data)
dfCopy=df
```

```python
duplicatedRow=df[df.duplicated()]
print("The duplcated Rows: \n",duplicatedRow)
```

10.Remove duplicate rows using drop_duplicates().

```python
df=df.drop_duplicates(ignore_index=True)
print("Removed duplicates:\n", df)
```

11. Identify and remove duplicate records based only on the Name column while
keeping the first occurrence.

```python
df=df.drop_duplicates(subset=['Name'], ignore_index=True, keep="first")
print("Removed Duplicates running on Name:\n",df)
```

12.Create a DataFrame containing both missing values and duplicate rows.

Clean the DataFrame by handling missing values and removing duplicates, then
display the final cleaned DataFrame.

```python
data = {
    "Name": ["Aarav", "Priya", "Rohan", "Priya", "Sneha", "Aarav"],
    "Age": [20, np.nan, 22, np.nan, 21, 20],
    "Marks": [85.0, 90.0, np.nan, 90.0, 78.0, 85.0],
    "City": ["Mumbai", "Delhi", "Pune", "Delhi", None, "Mumbai"],
}

df=pd.DataFrame(data)
dfCopy=df
```

```python
df['Age']=df['Age'].fillna((df['Age'].mean()).round(2))
df['Marks']=df['Marks'].fillna((df['Marks'].median()).round(2))
df['City']=df['City'].fillna("Unknown")

df=df.drop_duplicates(ignore_index=True)
print("Cleaned Data:\n",df)
```
