Q6. Given a = np.array([2, 4, 6]) and b = np.array([2.0, 4.0, 6.0]) - check both dtypes, then convert b

Q15. Given the marks array below - (a) extract all marks greater than 60; (b) extract marks between
50 and 80 using two conditions with & (remember the parentheses!).

marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

```python
marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])
ans1=marks[marks>60]
ans2=marks[(marks>=50)&(marks<=80)]

print(ans1)
print(ans2)
```
