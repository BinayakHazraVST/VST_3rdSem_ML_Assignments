Q1. A user types their city carelessly: " nEw dELhi ". Clean it and print exactly New Delhi. Also
validate that the city contains only alphabets and spaces — print Invalid city otherwise.

```python
city=input()

def validCity(city):
    city=city.strip().title()

    arr=city.split()

    for i in arr:
        if(i.isalpha()==False):
            return "Invalid City"
    
    return " ".join(arr)

print(validCity(city))
```

Q2. Prices arrive as one messy string: " 120, 45 , 300 ,80, 95 ". Convert it into a list of integers, then
print the total bill, the highest price, and the average price with 2 decimal places (use an f-string with :.2f).

```python
str=input()

def listInteger(str):
    str=str.strip()

    arr=[int(i) for i in str.split(",")]

    print("total price:",sum(arr))
    print("highest price:",max(arr))
    print(f"average price: {sum(arr)/len(arr):.2f}")

listInteger(str)
```

Q3. Validate a username entered by the user (e.g. " Rahul_123 "). Rules:
• clean the spaces and convert to lowercase
• must be alphanumeric (letters and digits only — no symbols)
• must start with an alphabet
• length must be between 5 and 12 characters
Print Valid username or Invalid username. ("Rahul_123" is invalid because of the underscore;
"rahul123" is valid.)

```python
user=input()

def validUsername(user):
    user=user.strip().lower()

    if(user[0].isalpha()==False):
        return "Invalid username"

    if(user.isalnum()==False):
        return "Invalid username"

    if(len(user)<5 or len(user)>12):
        return "Invalid username"

    return "Valid username"

print(validUsername(user))
```

Q4. For a given sentence, print: 
(a) the total number of words, 
(b) how many words end with a vowel, and 
(c) the longest word in the sentence.

```python
sen=input()

def sentenceParameter(sen):
    sen=sen.strip()
    arr=sen.split()

    for i in range(len(arr)):
        l=[]

        for j in range(len(arr[i])):
            if(arr[i][j].isalnum()==True):
                l.append(arr[i][j])

        arr[i]="".join(l)

    vowel="aeiouAEIOU"

    print("total words",len(arr))
    count=0

    for i in arr:
        if(i[-1] in vowel):
            count+=1

    print("words ending with vowel", count)

    maxL=0
    ans=0

    for i in arr:
        if maxL<=len(i):
            maxL=len(i)
            ans=i

    print("Longest word",ans)

sentenceParameter(sen)
    
```

Q5. Count the frequency of each character in a string, ignoring spaces and treating uppercase and lowercase
as the same letter. Example: "Aba c" → {'a': 2, 'b': 1, 'c': 1}.

```python
str=input()

def freq(str):
    hm={}
    str=str.lower().strip()

    for i in str:
        if(i==' '):
            continue

        else:
            hm[i]=hm.get(i,0)+1

    return hm

print(freq(str))
```

Q7. Invert the dictionary below so that values become keys. If two days share the same value, store both days
in a list:

```python
sales = {"mon": 200, "tue": 350, "wed": 200}

def invertKey(sales):
    inverted={}

    for day, value in sales.items():
        if(value not in inverted):
            inverted[value]=[day]
        else:
            inverted[value].append(day)

    return inverted

print(invertKey(sales))
```

Q8. Marks of two tests are stored in two dictionaries. Merge them into one dictionary — if a student appears in
both, sum the marks:

test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

```python
test1= {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

def mergeMarks(test1, test2):
    mark={}

    for name, marks in test1.items():
        if name not in mark:
            mark[name]=marks
        else:
            mark[name]+=marks

    for name, marks in test2.items():
        if name not in mark:
            mark[name]=marks
        else:
            mark[name]+=marks
    
    return mark

print(mergeMarks(test1,test2))
```

Q9. Sort the employees below by salary (highest first) using sorted() with key=lambda, and print each line
using an f-string in the format shown:

employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]

```python
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]

def salaryOrder(employees):
    ans=sorted(employees, key=lambda x:x[1], reverse=True)

    for data in ans:
        print(f"{data[0]} earns Rs. {data[1]}")

salaryOrder(employees)
```

Q10. Using the list of dictionaries below:

movies = [
 {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
 {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
 {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
 {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

• (a) Print the titles of all movies with rating > 8.1.
• (b) Extract all genres into a list (no duplicates).
• (c) Build a dict → {rating: [titles]} grouping movies by rating, then print it rating by rating using
f-strings, formatted nicely

```python
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

high_rated = [movie["title"] for movie in movies if movie["rating"] > 8.1]
for title in high_rated:
    print(title)

unique_genres = list({movie["genre"] for movie in movies})
print(unique_genres)

ratings_dict = {}
for movie in movies:
    rating = movie["rating"]
    title = movie["title"]
    
    if rating not in ratings_dict:
        ratings_dict[rating] = [title]
    else:
        ratings_dict[rating].append(title)

for rating, titles in ratings_dict.items():
    titles_str = ", ".join(titles)
    print(f"Rating {rating}: {titles_str}")
```
